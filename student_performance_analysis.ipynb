# Student Performance & Attendance Data Analysis using Pandas

# Project Tasks :
## Part 1: Data Loading & Cleaning :

## 1. Load the dataset

```python
import pandas as pd

Students_data = pd.read_csv('student_performance.csv')

Students_data
```

## 2. Show first 5 rows

```python
import pandas as pd 

Students_data.head()
```

## 3. Check basic information using df.info()


```python
import pandas as pd 

Students_data.info()
```

## 4. Check for missing values


```python
import pandas as pd 

Students_data.isnull().sum()
```

## 5. Replace missing numeric values with mean (if any)

```python
import pandas as pd 

Students_data['Math'] = Students_data['Math'].fillna((Students_data['Math'].mean()))
Students_data['Science'] = Students_data['Science'].fillna((Students_data['Science'].mean()))
Students_data['English'] = Students_data['English'].fillna((Students_data['English'].mean()))
Students_data['Attendance'] = Students_data['Attendance'].fillna((Students_data['Attendance'].mean()))


```

## 6. Remove duplicate rows if present

```python
import pandas as pd 

Students_data  = Students_data.drop_duplicates()

Students_data.info()
```

# Part 2: Data Analysis (Important) :
## 7. Display top 3 highest scorers in Math


```python
import pandas as pd 

Top_3_in_Math = Students_data.sort_values(by='Math',ascending=False).head(3)

Top_3_in_Math
```

## 8. Show all students whose Attendance < 80


```python
import pandas as pd 

Attendence_less_Than_80 = Students_data[Students_data['Attendance'] < 80 ]

Attendence_less_Than_80
```

## 9. Find average marks of each subject


```python
import pandas as pd 

Coloum  = ['Math','Science','English']

Average_Marks =  Students_data[Coloum].mean()

Average_Marks
```

## 10. Find the student with the highest total marks

```python
import pandas as pd 

Students_data['Total_Marks'] = Students_data[['Math','Science','English']].sum(axis=1)

Students_data['Total_Marks']

Highest_total_Marks  = Students_data.sort_values(by='Total_Marks',ascending=False)

print(Highest_total_Marks)

Highest_total_Marks.head(1)

```


## 11. Add a new column called Total (Math + Science + English)

```python
import pandas as pd 

Students_data['Total_Marks'] = Students_data[['Math','Science','English']].sum(axis=1)

Students_data
```

## 12. Add a new column Percentage

```python
import pandas as pd 

Students_data['Percentage'] = (Students_data['Total_Marks']/300)*100

Students_data
```

# Part 3: Grouping and Aggregation :


## 13. Find average marks city-wise


```python
import pandas as pd 

City_Wise_Grp = Students_data.groupby('City')['Total_Marks'].mean()

City_Wise_Grp
```

## 14. Count number of male and female students



```python
import pandas as pd 

Students_data.value_counts('Gender')
```

## 15. Find class-wise average attendance

```python
Average_Attendence = Students_data.groupby('Class')['Attendance'].mean()

Average_Attendence
```

# Part 4: Sorting and Filtering :


## 16. Sort students by Percentage (descending)


```python
import pandas as pd 

Sort_Students_By_Percentage = Students_data.sort_values(by='Percentage',ascending=False)

Sort_Students_By_Percentage
```

## 17. Show students who scored more than 90 in Science

```python
import pandas as pd 

Students_data[Students_data['Science'] > 90]
```


## 18. Show students from Delhi with attendance > 85

```python
import pandas as pd 

Students_data[(Students_data['City'] == 'Delhi') & (Students_data['Attendance'] > 85)]   

```

# Part 5: Pivot Table :


## 19. Create a pivot showing average marks for each gender


```python
import pandas as pd


Students_data["Average_Marks"] = Students_data[["Math",'Science','English']].mean(axis=1)

pd.pivot_table(Students_data,columns='Gender',values='Average_Marks')
```

## 20. Create pivot table of Attendance grouped by City & Gender

```python
import pandas as pd 

pd.pivot_table(Students_data,index='City',columns='Gender',values='Attendance')

```
